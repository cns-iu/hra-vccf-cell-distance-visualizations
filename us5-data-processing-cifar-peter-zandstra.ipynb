```python
# check if all dataframes have CelltypeName column
for i in range(len(df_list)):
    if "CelltypeName" not in df_list[i].columns:
        print(f"Dataframe {i} does not have CelltypeName column.")
    else:
        print(f"Dataframe {i} has CelltypeName column.")
```

```python
# Iterate through all dataframes. For each dataframe, get tissueID, x:x, y:y, and CelltypeName columns. 
# For each unique value in tissueID, get the corresponding x:x, y:y, and CelltypeName values. Create a new dataframe with these values.
# Save the new dataframe to a csv file in the output data path. NAme the csv file as tissueID.csv.
for i in range(len(df_list)):
    # Print the name of the dataframe.
    filename = os.listdir(INPUT_DATA_PATH)[i].split('.')[0]
    print(f"Processing dataframe {i}: {filename}")
    df = df_list[i]
    tissueID = df['tissueID'].unique()
    for j in range(len(tissueID)):
        # print the unique tissueID
        print(f"Processing tissueID {tissueID[j]}")
        tissueID_df = df[df['tissueID'] == tissueID[j]]
        tissueID_df = tissueID_df[['x:x', 'y:y', 'CelltypeName']]
        # Rename columns to x, y, and Cell Type.
        tissueID_df.columns = ['x', 'y', 'Cell Type']
        # Replace space in file name with underscore.
        filename = filename.replace(" ", "_")
        # Save to csv file.
        tissueID_df.to_csv(os.path.join(OUTPUT_DATA_PATH, f"{filename}--{tissueID[j]}.csv"), index=False)

# Print the names of all files in output data path.
print("Files in output data path:")
pprint(os.listdir(OUTPUT_DATA_PATH))
```

```python
# Visualize using Cell DIstance Exp
#Install and import external packages
# %pip install ipywidgets hra_jupyter_widgets

import os
import csv

import ipywidgets as widgets

# Import hra-jupyter-widgets. For documentation, please see https://github.com/x-atlas-consortia/hra-jupyter-widgets/blob/main/usage.ipynb
from hra_jupyter_widgets import CdeVisualization
```

```python

# Next, let's define a function that turns a DataFrame into a node list that can then be passed into the CdeVisualization or NodeDistVis widget
def make_node_list(df:pd.DataFrame, is_3d:bool = False):
  """Turn a DataFrame into a list of dicts for passing them into a HRA widget

  Args:
      df (pd.DataFrame): A DataFrame with cells
  """
  
  # If the df does not have a z-axis column, let's add one and set all cells to 0
  if not is_3d:
    df.loc[:, ('z')] = 0
  
  node_list = [{'x': row['x'], 'y': row['y'], 'z': row['z'], 'Cell Type': row['Cell Type']}
                 for index, row in df.iterrows()]

  return node_list
```

```python
# Read a random file from the output data path.
file = os.listdir(OUTPUT_DATA_PATH)[0]
df_nodes = pd.read_csv(os.path.join(OUTPUT_DATA_PATH, file))
node_list = make_node_list(df_nodes, False)
edge_list = [["Cell ID","Target ID", "X1", "Y1", "Z1", "X2", "Y2", "Z2"]] # this makes an empty edges list
```

```python
# Check if df_nodes has "ECs" or "Endothelial cells" in Cell Type column. Assign whichever exists to anchor_cell_type variable.
if "ECs" in df_nodes['Cell Type'].values:
    anchor_cell_type = "ECs"
elif "Endothelial cells" in df_nodes['Cell Type'].values:
    anchor_cell_type = "Endothelial cells"
else:
    anchor_cell_type = None
```

```python
# Finally, let's instantiate the CDEVisualization class with our node_list as parameter.
cde = CdeVisualization(
    node_target_selector=anchor_cell_type,
    max_edge_distance=200,
    nodes=node_list,
    node_target_key = "Cell Type",
)

# Display our new widget
display(cde)
```
